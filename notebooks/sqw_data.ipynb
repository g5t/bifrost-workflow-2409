```python
import scipp as sc
import numpy as np
```

```python
rng = np.random.default_rng(81273)
```

```python
nx, ny, nz, ne = 4, 4, 4, 5
nrun = 3
ndet = 2
nen = 2
nobs = nx * ny * nz * ne * nrun * ndet * nen

observations = sc.DataArray(
    sc.zeros(sizes={'obs': nobs}, with_variances=True),
    coords={
        'qx': sc.array(dims=['obs'], values=rng.uniform(-1, 1, nobs), unit='1/Å'),
        'qy': sc.array(dims=['obs'], values=rng.uniform(-1, 1, nobs), unit='1/Å'),
        'qz': sc.array(dims=['obs'], values=rng.uniform(-1, 1, nobs), unit='1/Å'),
        'e': sc.array(dims=['obs'], values=rng.uniform(0, 1, nobs), unit='meV'),
        'irun': sc.array(dims=['obs'], values=rng.integers(0, nrun, nobs), dtype='int64', unit=None),
        'idet': sc.array(dims=['obs'], values=rng.integers(0, ndet, nobs), dtype='int64', unit=None),
        'ien': sc.array(dims=['obs'], values=rng.integers(0, nen, nobs), dtype='int64', unit=None),
    }
)
binned = observations.bin(qx=nx, qy=ny, qz=nz, e=nen)
binned
```

```python
events = binned.bins.constituents['data']
pixels = np.vstack([
    *(events.coords[key].values
      for key in ('qx', 'qy', 'qz', 'e', 'irun', 'idet', 'ien')),
    events.data.values,
    events.data.variances,
])
pixels
```
